```python
import pandas as pd
import ast

df = pd.read_csv(r"C:\Users\hp\Documents\youtube_comments_cleann.csv")
df['text_tokens'] = df['text_tokens'].apply(ast.literal_eval)

print(df.shape)
```

```python
import pandas as pd

lexicon_pos = pd.read_csv("positive.tsv", sep='\t')
lexicon_neg = pd.read_csv("negative.tsv", sep='\t')

lexicon = pd.concat([lexicon_pos, lexicon_neg])
lexicon['weight'] = pd.to_numeric(lexicon['weight'], errors='coerce').fillna(0)
lexicon_dict = dict(zip(lexicon['word'], lexicon['weight']))

print (f"total kata yang ada di lexicon: {len(lexicon_dict)}")
print (f"contoh kata dan bobotnya: {list(lexicon_dict.items())[:20]}")
```

```python
def hitung_sentimen(tokens):
    score = 0
    for token in tokens:
        score += lexicon_dict.get(token, 0)
    return score
def label_sentimen(score):
    if score > 0:
        return 'positif'
    elif score < 0:
        return 'negatif'
    else:
        return 'netral'

df['sentimen_score'] = df['text_tokens'].apply(hitung_sentimen)
df['sentimen_label'] = df['sentimen_score'].apply(label_sentimen)

print(df[['text', 'text_tokens', 'sentimen_score', 'sentimen_label']].head(15).to_string())
print("\n Distribusi Label Sentimen:")
print(df['sentimen_label'].value_counts())
```

```python
df.to_csv(r"C:\Users\hp\Documents\youtube_comments_sentimen.csv", encoding='utf-8-sig')
print("Data anda telah dianalisis sentimennya dan disimpan ke file 'youtube_comments_sentimen.csv'")
print("\n Distribusi Label Final:")
print(df['sentimen_label'].value_counts())
```
