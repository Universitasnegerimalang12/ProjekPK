```python
import pandas as pd
df = pd.read_csv(r"C:\Users\hp\Documents\youtube_comments.csv")
print(df.head())
```

```python
import re

def bersihkan_teks(teks):
    teks = re.sub(r'@\w+', '', teks)
    teks = re.sub(r'#\w+', '', teks)
    teks = re.sub(r'[^\w\s]', '', teks)
    teks = re.sub(r'http\S+', '', teks)
    teks = re.sub(r'[^A-Za-z\s]', '', teks)
    teks = teks.lower() .strip()
    return teks
df['text_clean'] = df['text'].apply(bersihkan_teks)
print(df[['text', 'text_clean']].head(10).to_string())
```

```python
kamus_normalisai = {
    'gak': 'tidak', "yng ": "yang", 'mls': 'malas', 'klo': 'kalau', 'yg': 'yang', 'tdk': 'tidak', 'td': 'tadi', 'sklh': 'sekolah', 'gblk': 'bodoh', 'rmh': 'rumah', 'pur': 'purbaya', 'klo': 'kalo', 'g': 'tidak', 'ko': 'kenapa', 'kalo': 'kalau', 'duit': 'uang', 'rmh': 'rumah', 'gt': 'begitu', 'sma': 'sama', 'kl': 'kalau', 'trsrh': 'terserah', 'indo': 'indonesia', 'brk': 'buruk', 'nyantai': 'santai', 'koropsi': 'korupsi', 'puyeng': 'pusing', 'pake': 'pakai', 'maslah': 'masalah', 'represing': 'refrhesing', 'mng': 'mbg', 'owo': 'prabowo', 'wowo': 'prabowo','lbr': 'libur', 'spt': 'seperti', 'skr': 'sekarang', 'yg': 'yang', 'jg': 'juga', 'ko': 'kenapa', 'rmh': 'rumah', 'sya': 'saya', 'sy': 'saya', 'ancur': 'hancur', 'kesonoentahlah':'kesana entahlah', 'laper': 'lapar', 'ngirit': 'hemat', 'konoha': 'indonesia', 'rajia': 'razia', 'seyogyanya' : 'selogannya', 'keluyuran': 'keluyuran', 'diskolahan': 'di sekolah'}
def normalisasi(teks):
    kata_kata = teks.split()
    kata_kata= [kamus_normalisai.get(kata, kata) for kata in kata_kata]
    return ' '.join(kata_kata).strip()
df ['text_clean'] = df ['text_clean'].apply(normalisasi)
print(df[['text', 'text_clean']].head(15).to_string())
```

```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

Factory = StopWordRemoverFactory()
Stopword_Remover = Factory.create_stop_word_remover()

def hapus_stopword(teks):
    return Stopword_Remover.remove(teks)

df['text_clean'] = df['text_clean'].apply(hapus_stopword)
print(df[['text', 'text_clean']].head(15).to_string())
```

```python
df['text_tokens'] = df['text_clean'].apply(lambda x: x.split())
print(df[['text', 'text_clean', 'text_tokens']].head(15).to_string())
```
