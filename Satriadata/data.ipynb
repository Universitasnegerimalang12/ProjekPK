```python
from youtube_comment_downloader import YoutubeCommentDownloader,SORT_BY_POPULAR
import pandas as pd
import os

downloader = YoutubeCommentDownloader()
VIDEO_IDS = [
    "ICAjdI1gb3M",
    "GUhYf1s4C4o",
    "6p8ct9ZJSmo",
    ]

all_data = []
TOTAL_MAX= 300

for video_id in VIDEO_IDS:
    if len(all_data) >= TOTAL_MAX:
        break
    print(f"Downloading comments for video: {video_id}")
    try:
        comments = downloader.get_comments_from_url(f"https://www.youtube.com/watch?v={video_id}", sort_by=SORT_BY_POPULAR)
        for comment in comments:
            if len(all_data) >= TOTAL_MAX:
                break
            all_data.append({
                "video_id": video_id,
                "text": comment["text"],
                "author": comment["author"],
                "likes": comment["votes"],
                "timestamp": comment["time"],
            })
        print(f"Downloaded {len(all_data)} comments so far...")
    except Exception as e:
        print(f"Error downloading comments for video {video_id}: {e}")
        
#simpan
df = pd.DataFrame(all_data)
df.drop_duplicates(subset=["text"], inplace=True)

desktop = os.path.join(os.path.expanduser("~"), "Documents")
output_path = os.path.join(desktop, "youtube_comments.csv")
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Saved {len(df)} unique comments to {output_path}")
print(f"file:{output_path} created successfully.")
print(df.head())
```
